```python
from mcda.matrices import *
from mcda.relations import *
from mcda.scales import *
from mcda.outranking.srmp import *
from pandas import Series
import matplotlib.pyplot as plt
```

```python
alternatives = ["A", "B", "C", "D"]

criteria = ["Confort", "Prix (€)", "Design"]

scales = {
    "Confort": QualitativeScale(
        Series([1, 2, 3, 4, 5], index=["Exigu", " ", "Satisfaisant", "  ", "Spacieux"]), preference_direction=MAX
    ),
    "Prix (€)": QuantitativeScale(22000, 26000, preference_direction=MIN),
    "Design": QualitativeScale(
        Series([1, 2, 3, 4, 5], index=["Moche", " ", "Acceptable", "  ", "Élégant"]), preference_direction=MAX
    ),
}

dataset = PerformanceTable(
    [
        ["Satisfaisant", 22620, "Moche"],
        ["Spacieux", 24700, "Acceptable"],
        ["Exigu", 25740, "Élégant"],
        ["Satisfaisant", 23400, "Acceptable"],
    ],
    scales,
    alternatives,
    criteria,
)
```

```python
SRMP.plot_input_data(dataset, annotations=True, annotations_alpha=1, figsize=(8, 6))
plt.savefig("alternatives.pdf")
```

```python
profiles = PerformanceTable(
    [
        [" ", 25000, " "],
        ["  ", 24000, "  "],
    ],
    scales,
    criteria=criteria,
)
```

```python
SRMP.plot_input_data(dataset, profiles=profiles, annotations=True, annotations_alpha=0.5, figsize=(8, 6))
plt.savefig("profiles.pdf")
```

```python
import tikzplotlib
SRMP.plot_input_data(
    dataset,
    profiles=profiles,
    lexicographic_order=[0, 1],
    annotations_alpha=1,
    figsize=(8, 6),
)
plt.savefig("lex_order.pdf")
tikzplotlib.save("mytikz.tex", standalone=True)
```

```python
criteria_weights = {
    "Confort": 0.4,
    "Prix (€)": 0.5,
    "Design": 0.1,
}
```

```python
SRMP.plot_input_data(
    dataset,
    criteria_weights=criteria_weights,
    profiles=profiles,
    lexicographic_order=[0, 1],
    annotations_alpha=1,
    figsize=(8, 6),
)
plt.savefig("lex_order.pdf")
```

```python
SRMP.plot_input_data(
    dataset,
    criteria_weights=criteria_weights,
    profiles=profiles,
    # lexicographic_order=[1, 0],
    annotations_alpha=1,
    figsize=(4, 3),
)
plt.savefig("weights.pdf")
```
